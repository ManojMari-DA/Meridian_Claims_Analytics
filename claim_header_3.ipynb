```python
columns=['total_billed','total_allowed','total_paid']
```

```python
for col in columns:

    Q1 = ch3[col].quantile(0.25)
    Q3 = ch3[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    print("\nColumn:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
```

```python
total_billed = ch3[
    (ch3['total_billed'] < -200.9699999999999) |
    (ch3['total_billed'] >1490.6299999999999 )
]

total_allowed = ch3[
    (ch3['total_allowed'] <-471.05 ) |
    (ch3['total_allowed'] >911.27 )
]

total_paid= ch3[
    (ch3['total_paid'] <-374.7950000000001 ) |
    (ch3['total_paid'] > 729.085 )
]

print("total_billed_anomalies:", len(total_billed))
print("total_allowed_anomalies:", len(total_allowed))
print("total_paid_anomalies:", len(total_paid))
```

```python
ch3['total_billed_anomoly'] = (
    (ch3['total_billed'] <-200.9699999999999 ) |
    (ch3['total_billed'] > 1490.6299999999999 )
)

ch3['total_allowed_anomaly'] = (
    (ch3['total_allowed'] < -471.05 ) |
    (ch3['total_allowed'] >911.27 )
)

ch3['total_paid_anomaly'] = (
    (ch3['total_paid'] <-374.7950000000001 ) |
    (ch3['total_paid'] >729.085 )

)
```

```python
all_anomalies = ch3[
    (ch3['total_billed_anomoly']) |
    (ch3['total_allowed_anomaly']) |
    (ch3['total_paid_anomaly'])
]

print("Total Unique Anomaly Records:", len(all_anomalies))
```
