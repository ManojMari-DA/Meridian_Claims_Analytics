```python
columns=['billed_amount','allowed_amount','paid_amount']
```

```python
for col in columns:

    Q1 = cl2[col].quantile(0.25)
    Q3 = cl2[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    print("\nColumn:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
```

```python
# Billed Amount
billed_anomalies = cl2[
    (cl2['billed_amount'] < -632.835) |
    (cl2['billed_amount'] >1267.285 )
]

# Allowed Amount
allowed_anomalies = cl2[
    (cl2['allowed_amount'] <-390.615 ) |
    (cl2['allowed_amount'] > 651.0250000000001)
]

# Paid Amount
paid_anomalies = cl2[
    (cl2['paid_amount'] <-290.34749999999997 ) |
    (cl2['paid_amount'] > 483.91249999999997)
]

print("Billed Anomalies:", len(billed_anomalies))
print("Allowed Anomalies:", len(allowed_anomalies))
print("Paid Anomalies:", len(paid_anomalies))
```

```python
cl2['billed_anomaly'] = (
    (cl2['billed_amount'] < -632.835) |
    (cl2['billed_amount'] > 1267.285)
)

cl2['allowed_anomaly'] = (
    (cl2['allowed_amount'] <-390.615  ) |
    (cl2['allowed_amount'] >651.0250000000001 )
)

cl2['paid_anomaly'] = (
    (cl2['paid_amount'] < -290.34749999999997 ) |
    (cl2['paid_amount'] > 483.91249999999997)
)
```

```python
all_anomalies = cl2[
    (cl2['billed_anomaly']) |
    (cl2['allowed_anomaly']) |
    (cl2['paid_anomaly'])
]

print("Total Unique Anomaly Records:", len(all_anomalies))
```
