```python
ch2.rename(columns={
    "mbr_id": "member_id",
    "svc_from_dt": "service_from_date",
    "svc_to_dt": "service_to_date",
    "allowed_amt": "total_allowed",
    "paid_amt": "total_paid"
}, inplace=True)
```

```python
columns=['total_billed','total_allowed','total_paid']
```

```python
for col in columns:

    Q1 = ch2[col].quantile(0.25)
    Q3 = ch2[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    print("\nColumn:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
```

```python
total_billed = ch2[
    (ch2['total_billed'] < -184.1175) |
    (ch2['total_billed'] > 1456.1625)
]

total_allowed = ch2[
    (ch2['total_allowed'] < -563.09625) |
    (ch2['total_allowed'] >938.4937500000001 )
]

total_paid= ch2[
    (ch2['total_paid'] < -445.42875 ) |
    (ch2['total_paid'] > 742.3812499999999)
]

print("total_billed_anomalies:", len(total_billed))
print("total_allowed_anomalies:", len(total_allowed))
print("total_paid_anomalies:", len(total_paid))

```

```python
ch2['total_billed_anomoly'] = (
    (ch2['total_billed'] <-184.1175 ) |
    (ch2['total_billed'] > 1456.1625)
)

ch2['total_allowed_anomaly'] = (
    (ch2['total_allowed'] < -563.09625) |
    (ch2['total_allowed'] > 938.4937500000001)
)

ch2['total_paid_anomaly'] = (
    (ch2['total_paid'] < -445.42875 ) |
    (ch2['total_paid'] > 742.3812499999999)
)
```

```python
all_anomalies = ch2[
    (ch2['total_billed_anomoly']) |
    (ch2['total_allowed_anomaly']) |
    (ch2['total_paid_anomaly'])
]

print("Total Unique Anomaly Records:", len(all_anomalies))
```
