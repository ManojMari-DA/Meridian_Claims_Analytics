```python
columns=['premium_amount']
```

```python
for col in columns:

    Q1 = p3[col].quantile(0.25)
    Q3 = p3[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    print("\nColumn:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
```

```python
premium_amount = p3[
    (p3['premium_amount'] < 110.0) |
    (p3['premium_amount'] > 830.0)
]
print("premium_amount_anomalies:", len(premium_amount))
```
