```python
columns=['billed_amount','allowed_amount','paid_amount']
```

```python
for col in columns:

    Q1 = cl3[col].quantile(0.25)
    Q3 = cl3[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    print("\nColumn:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
```

```python
# Billed Amount
billed_anomalies = cl3[
    (cl3['billed_amount'] <-672.8675000000002 ) |
    (cl3['billed_amount'] > 1328.8725000000002)
]

# Allowed Amount
allowed_anomalies = cl3[
    (cl3['allowed_amount'] <-374.98875000000004 ) |
    (cl3['allowed_amount'] >655.4212500000001 )
]

# Paid Amount
paid_anomalies = cl3[
    (cl3['paid_amount'] <-288.565 ) |
    (cl3['paid_amount'] >505.07500000000005 )
]

print("Billed Anomalies:", len(billed_anomalies))
print("Allowed Anomalies:", len(allowed_anomalies))
print("Paid Anomalies:", len(paid_anomalies))
```

```python
cl3['billed_anomaly'] = (
    (cl3['billed_amount'] < -672.8675000000002 ) |
    (cl3['billed_amount'] > 1328.8725000000002)
)

cl3['allowed_anomaly'] = (
    (cl3['allowed_amount'] < -374.98875000000004) |
    (cl3['allowed_amount'] >655.4212500000001 )
)

cl3['paid_anomaly'] = (
    (cl3['paid_amount'] <-288.565 ) |
    (cl3['paid_amount'] > 505.07500000000005)
)
```

```python
all_anomalies = cl3[
    (cl3['billed_anomaly']) |
    (cl3['allowed_anomaly']) |
    (cl3['paid_anomaly'])
]

print("Total Unique Anomaly Records:", len(all_anomalies))
```
