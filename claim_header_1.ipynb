```python
columns=['total_billed','total_allowed','total_paid']
```

```python
for col in columns:

    Q1 = ch1[col].quantile(0.25)
    Q3 = ch1[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    print("\nColumn:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
```

```python
total_billed = ch1[
    (ch1['total_billed'] < -237.91000) |
    (ch1['total_billed'] > 1579.77)
]

total_allowed = ch1[
    (ch1['total_allowed'] <-433.68499 ) |
    (ch1['total_allowed'] > 909.375)
]

total_paid= ch1[
    (ch1['total_paid'] <  -345.27000) |
    (ch1['total_paid'] > 723.17000 )
]

print("total_billed_anomalies:", len(total_billed))
print("total_allowed_anomalies:", len(total_allowed))
print("total_paid_anomalies:", len(total_paid))
```

```python
ch1['total_billed_anomoly'] = (
    (ch1['total_billed'] < -237.91000) |
    (ch1['total_billed'] > 1579.77)
)

ch1['total_allowed_anomaly'] = (
    (ch1['total_allowed'] < -433.68499) |
    (ch1['total_allowed'] >909.375 )
)

ch1['total_paid_anomaly'] = (
    (ch1['total_paid'] < -345.27000) |
    (ch1['total_paid'] >723.17000 )
)
```

```python
all_anomalies = ch1[
    (ch1['total_billed_anomoly']) |
    (ch1['total_allowed_anomaly']) |
    (ch1['total_paid_anomaly'])
]

print("Total Unique Anomaly Records:", len(all_anomalies))
```
