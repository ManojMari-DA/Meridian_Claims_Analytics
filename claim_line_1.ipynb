```python
columns=['billed_amount','allowed_amount','paid_amount']
```

```python
for col in columns:

    Q1 = cl1[col].quantile(0.25)
    Q3 = cl1[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    print("\nColumn:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower)
    print("Upper Bound:", upper)
```

```python
# Billed Amount
billed_anomalies = cl1[
    (cl1['billed_amount'] < -614.54625) |
    (cl1['billed_amount'] > 1240.20375)
]

# Allowed Amount
allowed_anomalies = cl1[
    (cl1['allowed_amount'] < -334.60625) |
    (cl1['allowed_amount'] > 617.62375)
]

# Paid Amount
paid_anomalies = cl1[
    (cl1['paid_amount'] < -245.21) |
    (cl1['paid_amount'] > 456.39)
]

print("Billed Anomalies:", len(billed_anomalies))
print("Allowed Anomalies:", len(allowed_anomalies))
print("Paid Anomalies:", len(paid_anomalies))
```

```python
cl1['billed_anomaly'] = (
    (cl1['billed_amount'] < -614.54625) |
    (cl1['billed_amount'] > 1240.20375)
)

cl1['allowed_anomaly'] = (
    (cl1['allowed_amount'] < -334.60625) |
    (cl1['allowed_amount'] > 617.62375)
)

cl1['paid_anomaly'] = (
    (cl1['paid_amount'] < -245.21) |
    (cl1['paid_amount'] > 456.39)
)
```

```python
all_anomalies = cl1[
    (cl1['billed_anomaly']) |
    (cl1['allowed_anomaly']) |
    (cl1['paid_anomaly'])
]

print("Total Unique Anomaly Records:", len(all_anomalies))
```
